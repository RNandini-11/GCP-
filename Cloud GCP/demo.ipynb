#### **Create buckets:**

```python
from google.cloud import storage

client = storage.Client()
bucket = client.create_bucket("gcp-nandini-1103")
print("Bucket created successfully...!")
```

**List Buckets:**

```python
from google.cloud import storage

client = storage.Client()

for bucket in client.list_buckets():
    print(bucket.name)
```

**Upload object:**

```python
from google.cloud import storage

client = storage.Client()
bucket = client.bucket("gcp-nandini-1103")
blob = bucket.blob("notes.txt")
blob.upload_from_filename(r"D:\nandini\Cloud GCP\notes.txt")
print("Blob uploaded successfully...!")
```

**Copy object:**

```python
from google.cloud import storage

client = storage.Client()
bucket = client.bucket("gcp-nandini-1103")
source_blob = bucket.blob("notes.txt")
bucket.copy_blob(
    source_blob,
    bucket,
    "nandini.txt"
)
print("Object copied successfully...!")
```

**Move object:**

```python
from google.cloud import storage

client = storage.Client()

bucket = client.bucket("gcp-nandini-1103")
source_blob = bucket.blob("notes.txt")
bucket.copy_blob(
    source_blob,
    bucket,
    "Archive/notes.txt"
)
source_blob.delete()
print("Object moved successfully...!")
```

**Delete object:**

```python
from google.cloud import storage

client = storage.Client()
bucket = client.bucket("gcp-nandini-1103")
blob = bucket.blob("nandini.txt")
blob.delete()
print("Object deleted successfully...!")
```
