```python
# create bucket
from google.cloud import storage

client = storage.Client()
bucket_name = "gcp-nandini-1103"
bucket = client.create_bucket(bucket_name)


print(f"{bucket_name} - bucket created successfully..!")
```

```python
# upload object

bucket = client.bucket("gcp-nandini-1103")
blob = bucket.blob("notes.txt")
blob.upload_from_filename(r"D:\nandini\demo\notes.txt")
print("notes.txt file uploaded successfully..!")
```

```python
# copy object

bucket = client.bucket("gcp-nandini-1103")
source_blob = bucket.blob("notes.txt")
bucket.copy_blob(
    source_blob,
    bucket,
    "nandini.txt"
)
print("Object copied successfully...!")
```

```python
# move object

bucket = client.bucket("gcp-nandini-1103")
source_blob = bucket.blob("notes.txt")
bucket.copy_blob(
    source_blob,
    bucket,
    "Archive/notes.txt"
)
source_blob.delete()
print("Object copied successfully...!")
```

```python
# delete object

bucket = client.bucket("gcp-nandini-1103")
blob = bucket.blob("notes.txt")
blob.delete()
print("Object deleted successfully...!")
```

```python
bucket = client.bucket("gcp-nandini-1103")
blob = bucket.blob("Archive/notes.txt")
blob.delete()
print("Object deleted successfully...!")
```

```python
bucket = client.bucket("gcp-nandini-1103")
blob = bucket.blob("nandini.txt")
blob.delete()
print("Object deleted successfully...!")
```

```python
# delete bucket

bucket = client.get_bucket("gcp-nandini-1103")
bucket.delete()
print("Bucket deleted successfully...!")
```
